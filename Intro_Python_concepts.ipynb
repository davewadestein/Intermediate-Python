# Important things to know about Python
* dynamically typed
  * no type defined along with variables
  * Python does not prevent us from putting some data of another type in that variable
* scalars vs. containers
  * scalar = single value (e.g., int, float, bool)
  * container = thing that holds 0+ other things
    * list, str, dict, tuple, set
    * ordered sequences: list, str, tuple
    * unordered (bag) of values: set, dict
* mutable vs. immutable types
  * (im)mutable = (not) changeable
  * mutable: list, set, dict
  * immutable: str, tuple, frozenset
* function?
  * reusable snippet of code
  * some are builtin
  * like an "appliance"
    * coffee machine
      * input: water, coffee beans/groud
      * output: drinkable coffee
    * blender
      * input: yogurt, fruit
      * output: smoothie
      * action: chops up food, mixes food/liquid
    * garbage disposer
      * input: food waste
      * output: none
    * television
      * input: none
      * output: entertainment
    * heater
      * input: none
      * output: heat, comfort
  * how are functions called?
    * e.g., __`print('call the print function')`__
    * we name it, add parens, and pass parameter or arguments or data

* method?

* module = Python's term for a library–technically, a module is a file of Python code containing:
  * functions
  * classes (tbd)
  * methods (tbd)
* list
  * mutable
  * typically used to hold homogeneous data, i.e., like a column of a spreadsheet
* tuple
  * immutable
  * any comma-separated sequence of value is a tuple
  * typically for heterogeneous data, i.e., a ROW of a spreadsheet
* the built-in functions are general in nature, that is to say they can be called with many different types
  * built-in functions do not change the objects that are passed to them
  * if you want to change an object in Python you must call/invoke/apply a method on/to that object (e.g., __`list.append`__)
    * not all methods change the objects they are called/invoked/applied on/to
* list vs. array?
* "good artists borrow, great artist steal" –Picasso
* "Pythonic"
  * writing our code in a way that experienced Python programmers expect to see it
  * using idioms that other programmers expect
    * __`container[-1]`__ always means the last item in the container
    * listcomps
  * truthy/falsy
    * non-zero values (including negatives) are considered True in a Boolean context
      * 0 and 0.0 are considered False in a Boolean context
      * None is considered False...
    * non-empty containers are considered True in a Boolean context
      * empty containers are consider False " " " "
* reading code
  * we read code 10x more than we write code
  * therefore we may encounter constructs in other people's code that we don't use
* Hal Abelson: "Programs must be written for people to read, and only incidentally for machines to execute."
  * Eagleson's Law: "Any code you wrote longer than 6 months ago might as well have been written by someone else"
* everything in Python is an object
  * everything lives in memory and can be inspected
    * __`id()`__ gives us the address/identity
    * __`dir()`__ cracks open an object and lets us see what's inside it
* is Python pass-by-value or pass-by-reference?
  * pass-by-value: when you call a function, the object gets copied
    * C is completely pass-by-value
  * pass-by-reference: when you call a function, the function gets access to the original argument and can change if it wants to
    * C++ is pass-by-value or pass-by-reference
  * Python is actually "pass by assignment"

```python
import random # bring a module into Python's memory so we can use it

random.seed(123)

# nums is a/becomes a list of 100 random integers from 1 to 10
nums = [random.randint(1, 10) for _ in range(15)]
print(nums)
```

```python
# turn the above into a standard loop
nums = []
random.seed(3)

for _ in range(15):
    nums.append(random.randint(1, 10))

print(nums)
```

```python
# filter the list of numbers into a list of even numbers
even_numbers = []

for num in nums: # for each number in the nums list
    if num % 2 == 0: # if remainder when dividing by 2 is 0 (i.e., even)
        even_numbers.append(num)

even_numbers
```

```python
# [(transformed) values     where do the values come from?
#                            (filter to choose which items)]
newnums = [num for num in nums
                  if num % 2 == 0]
newnums
```

```python
# Use a list comprehension to create a list of the integers from 1 to 100 which are not divisible by 5
# first, notice than if we wanted a list of numbers from 1 to 100,
# we'd use list(range(1, 101)) ... no need for a listcomp
# but we want to filter out every 5th one, so we need a listcomp
no_divis_by_5 = [num for num in range(1, 101)
                        if num % 5 != 0] # Pythonic
print(no_divis_by_5)
```

```python
def product(*args):
    """Multiply together all of the arguments and return result."""
    # remember that we'll receive a tuple here and we can iterate over it
    # if no arguments were received (empty tuple), we want the result to be 1
    result = 1 # start with 1, and multiply all args together...
    
    for arg in args: # note that for loop won't run if args is an empty tuple
        result *= arg

    return result
```

```python
class StrList(list, dict):
    """Won't work due to incompatibilities in structure
    str: immutable
    list: mutable

    Multiple inheritance works best when you wish to combine behaviors of parent classes
    """
    pass
```

```python
class Loggable:
    def log(self):
        """substitute in your mind some useful behavior"""
        print('log messages')
```

```python
class Saveable: # implicitly this class inherits from "object"
    """substitute in your mind some useful behavior"""
    def save(self):
        print('save info to disk')
```

```python
class Document(Loggable, Saveable): # mix-in
    """A documemt inherits all the chocolately goodness from Loggable and Saveable"""
    """the essence of multiple inheritance is to inherit both behaviors from the parent"""
    def __init__(self):
        print('in __init__')
```

```python
d = Document(2) # what function runs here?
```

```python
d.log()
```
