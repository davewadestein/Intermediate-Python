# The __`collections`__ module
* the __`collections`__ module contains a bunch of useful types which are derived from (_read: inherited from_) some of the built-in types we're already familiar with

## Ordered Dictionaries
* prior to Python 3.6, items inserted into dictionaries did not retain their insertion order
* no longer an issue
* althought OrderedDict does have a new method that dicts don't have

```python
from collections import OrderedDict

d = OrderedDict()
d['one'] = 3
d['two'] = 6
d['three'] = 0
d # vs. print(d)
```

```python
d.move_to_end('one')
print(d)
```

```python
d.move_to_end('one', False)
print(d)
```

## Default Dictionaries
* suppose we need a default value for any key which does not exist in the dictionary
 * we can
     * use the __`get()`__ function
     * use __`setdefault()`__
     * use the __`in`__ operator)
* ...or we can use a __`defaultdict`__

```python
# Using .get()

def count_letters(word):
    """Returns a dict how many times the letter appeared in the word/str passed in"""
    count = {}
    for ltr in word:
        count[ltr] = count.get(ltr, 0) + 1
    return count

count_letters('antidisestablishmentarianism')
```

```python
# using a default dict instead
from collections import defaultdict

def count_letters(word):
    """Returns} a dict how many times the letter appeared in the word/str passed in"""
    # When creating a defaultdict, the passed argument dictates what the
    # default value will be (int = 0, str = "", list = [])
    count = defaultdict(int)
    
    for ltr in word:
        count[ltr] += 1
    return count

count_letters('one two three four two three three')
```

# Deque
* double ended queue
* pronounced "deck"

```python
from collections import deque
dq = deque(range(10), maxlen=10) # maxlen is optional
print(dq)
```

```python
dq.rotate(3) # +n takes items from right, prepends to left, vice versa for -n
print(dq)
```

```python
dq.rotate(-4)
print(dq)
```

```python
dq.appendleft('a') # appending to full deque discards item(s) from other end
print(dq)
```

```python
dq.extend('bcd')
print(dq)
```

```python
dq.extendleft((-1, -2, -3))
print(dq)
```

```python
dq.pop() # same as list
```

```python
dq.popleft() # specific to deque, as is rotate()
```

```python
print(dq)
dq.remove(3) # same as list
print(dq)
```

```python
dq.reverse()
print(dq)
```

```python
dq.append(0)
dq
```

## Named Tuples
* tuples are quite handy, but they are missing a key feature when using them as records–sometimes we want to name the fields
 * more efficient (i.e., less memory) than dictionaries because instances don't need to contain the keys themselves, as dictionaries do, just the values
* __`namedtuple()`__ returns not an individual object but a new class, customized for the given names

```python
from collections import namedtuple

# Create a namedtuple called "Point"
# the first arg 'Point' indicates what the classes internal name will be
# the second arg is the desired attribute names as an iterable of strings or a
# single space/comma-delimited string
Point = namedtuple('Point', 'x y')

# Note that the above does not create an object, but rather, it creates a new TYPE
# (technically, namedtuple is a "class factory")

point1 = Point(1, 3)
print(point1, type(point1), sep='\n')
```

```python
point2 = Point(-3, -2)
print(point2)
```

```python
# what we would do if it were just a plain old tuple
print(point1[0], point1[1]) 
```

```python
# but we can, if we wish, do this!
print(point1.x, point1.y)
```

```python
from collections import namedtuple

City = namedtuple('City', 'name country population coordinates')
tokyo = City('Tokyo', 'JP', 36.933, (35.689722, 139.691667))
print(tokyo)
```

```python
print(tokyo.population) # Prefer to use attribute or field names
print(tokyo.coordinates)
print(tokyo[1]) # use indexing if we wish
```

```python
type(tokyo)
```

```python
type(City)
```

```python
for field in City._fields: # tuple containing field names
    print(field)
```

```python
LatLong = namedtuple('LatLong', 'lat long')
# the below is a regular tuple (which contains a namedtuple)
delhi_data = 'Delhi NCR', 'IN', 21.935, LatLong(28.613889, 77.2098889)
```

```python
# the _make() method creates a namedtype from an iterable of fields
delhi = City._make(delhi_data)
delhi
```

```python
# but we can get the same behavior using Python's "unpack" operator
delhi2 = City(*delhi_data)
delhi2
```

```python
delhi == delhi2
```

```python
d = delhi._asdict() # returns an OrderedDict built from named tuple
print(d)
```

## Counters
* __`dict`__ subclass for counting things
* unordered collection where things being counted are `dict` keys and the counts are `dict` values
* __`Counters`__ can have negative values

```python
from collections import Counter
c = Counter()
c
```

```python
c = Counter('antidisestablishmentarianism')
c
```

```python
c.update('establish' * 10)
c
```

```python
c = Counter({'red': 5, 'blue': -1})
c
```

```python
c = Counter(foo=1, bar=2)
c
```

```python
c = Counter(red=6, blue=5, green=3, pink=1, yellow=-3)
c.elements() # returns an iterator
```

```python
for thing in c.elements(): # cf. list(...)
    print(thing, end=' ')
```

```python
c.most_common(3) # returns the n most common elements
```

```python
d = Counter(fuschia=3, pink=0, red=3, blue=5, green=2)
c.subtract(d) # preserves negative values
c
```

```python
c.items() # remember that under the hood, this is a dict
```

```python
+c # generates new Counter, discarding 0s or negatives
```

```python
c = +c
c
```

```python
c = Counter(red=6, blue=-5, green=3, pink=1, yellow=-3)
c = -c # discard positives and multiply remaining negatives by -1
c
```

```python
d = Counter(red=6, yellow=7, green=9)
c.update(d)
c
```

```python
c = Counter(a=3, b=1, c=4)
d = Counter(a=1, b=2, c=5)
c + d
```

```python
c - d
```

```python
print(c, d, sep='\n')
```
