## Object-Oriented (OO) Programming and Python Classes
* programming paradigm based on the concept of "objects" rather than "actions"
  * objects may contain data, i.e., fields–often called *attributes*
  * objects may contain code, i.e., functions–often called *methods*
  * an object's methods can access and (often) modify the data fields of the object with which they are associated (objects have a notion of *this* or *self*)
* OO programs are created from objects that interact with one another
* Python, like most popular OO programming languages, is class-based
  * i.e., objects are instances of classes, which also determine their type
  * in fact, it's helpful to think of a class as a type–they are synonymous

## What a Class Represents

A class is a way to **group related data and behavior together**.

In other words, a class can represent:

- a real-world thing, such as a `BankAccount`, `Car`, or `Customer`
- a logical concept, such as a `Connection`, `Logger`, or `Report`

A class gives us a way to say:

> "These pieces of data and these operations belong together."


```python
# This class represents a bank account. I like this example because everyone
# has a bank account and therefore they can relate to the idea of a bank account.
    
class BankAccount:
    """
    A class which represents, perhaps unsurpisingly, a bank account.
    
    Attributes:
        owner (str): Owner of the bank account.
        balance (float): Owner's balance.  
    
    """
    def __init__(self, owner, balance): # ignore "self" for the moment
        """Initializer method. Params are the data neeed to creat a bank account."""
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

    def show_balance(self):
        print(f"{self.owner} has ${self.balance}")
```

This class represents something meaningful and familiar:

- a bank account has an **owner**
- a bank account has a **balance**
- a bank account can **deposit** money
- a bank account can **withdraw** money


## Objects and Instances

A **class** is like a blueprint.

An **object** (or **instance**) is a specific thing created from that blueprint.


```python
account1 = BankAccount("Avery", 100) # why not 3 parameters?
account2 = BankAccount("Jordan", 250)
```

```python
print(account1)
print(account2)
```

Both `account1` and `account2` are instances of the same class, but they hold different data.


```python
print(account1.owner, account1.balance)
print(account2.owner, account2.balance)
```

- `BankAccount` is the class
- `account1` and `account2` are instances

(Each instance has its own state.)

## Attributes and Methods

In OOP, a class usually contains:

- **attributes**: data stored on the object
- **methods**: functions attached to the object

In our `BankAccount` example:

- `owner` and `balance` are attributes
- `deposit()`, `withdraw()`, and `show_balance()` are methods


```python
account1.show_balance()
account1.deposit(50)
account1.show_balance()
account1.withdraw(20)
account1.show_balance()
```

* The __`self`__ parameter refers to the specific instance the method is being called on.


```python
class Customer:
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def describe(self):
        print(f"{self.name} can be reached at {self.email}")

    def update_email(self, new_email):
        self.email = new_email
```

```python
customer = Customer("Sam Lee", "sam@example.com")
customer.describe()
customer.update_email("sam.lee@example.com")
customer.describe()
```

Here, the methods operate on the object's data.

That is one of the main ideas of OOP:

> the data and the behavior that acts on that data live together


## Why Use Classes Instead of Dicts or Separate Functions?

Python gives us many ways to structure code. Sometimes a dictionary is enough. Sometimes a few functions are enough.

So why introduce classes at all?


```python
customer = {
    "name": "Sam Lee",
    "email": "sam@example.com"
}

def describe_customer(c):
    print(f"{c['name']} can be reached at {c['email']}")

def update_customer_email(c, new_email):
    c["email"] = new_email

describe_customer(customer)
update_customer_email(customer, "sam.lee@example.com")
describe_customer(customer)
```

This works. So what is the issue?

A dictionary + separate functions can become harder to manage when:

- the structure grows
- behavior becomes more complex
- many related functions exist
- you want a clearer model of the concept

Compare that to the class-based version.


```python
class Customer:
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def describe(self):
        print(f"{self.name} can be reached at {self.email}")

    def update_email(self, new_email):
        self.email = new_email

customer = Customer("Sam Lee", "sam@example.com")
customer.describe()
customer.update_email("sam.lee@example.com")
customer.describe()
```

* The class version makes the relationship between the data and behavior more explicit and can improve
  - organization
  - readability
  - maintainability


## Rule of Thumb

Use a class when:

- you have data and behavior that belong together
- you expect multiple similar objects
- you want a clearer structure than loose dictionaries and functions

Do **not** use a class just because OOP exists.


## Structure and Organization

Classes can help organize code by giving related ideas a home

Without a class, code often spreads across:

- dictionaries
- helper functions
- global state

With a class, related logic can be grouped together


```python
class Car:
    def __init__(self, make, model, fuel_level):
        self.make = make
        self.model = model
        self.fuel_level = fuel_level

    def drive(self, miles):
        self.fuel_level -= miles * 0.05

    def refuel(self, gallons):
        self.fuel_level += gallons

    def status(self):
        print(f"{self.make} {self.model} has {self.fuel_level:.1f} gallons of fuel")
```

```python
car = Car("Toyota", "Camry", 10)
car.status()
car.drive(40)
car.status()
car.refuel(3)
car.status()
```

The __`Car`__ class groups together:

- the data (__`make`__, __`model`__, __`fuel_level`__)
- the related operations (__`drive`__, __`refuel`__, __`status`__)

That is often easier to reason about than several disconnected functions


## Readability and Maintainability

A good class can make code easier to read because it reflects the real shape of the problem

For example, compare:

- __`customer.email`__
- __`customer.update_email('new@example.com')`__

...to a pile of unrelated helper functions

Classes can help answer questions like:

- What does this object know?
- What can this object do?
- Where should this new logic go?


At the same time, classes are **not automatically better**.

A poorly designed class can hurt readability just as much as poorly organized procedural code

Important goals:

- keep classes focused
- give them clear responsibilities
- avoid turning one class into a giant dumping ground

## Inheritance
* a class may "inherit" from another class (or more than one class)
  * the "child" class inherits "all the chocolately goodness" of the parent, but then either
    * adds new functionality that didn't exist in the parent ("extends" the parent)
    * or alters the functionality in the child compared to the parent ("overrides" the parent)
    * ...or both

```python
class BetterList(list): # BetterList inherits from the built-in list class
    """Add a ".removeall()" method"""
    
    def removeall(self, item):
        """Remove *all* instances of item from the list"""
        while item in self:
            self.remove(item)
```

```python
test_list = BetterList([1] * 10 + [2])
test_list, type(test_list)
```

```python
test_list.removeall(1)
test_list
```
